```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

# Padding without Strides

```python
model1 = Sequential()

model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu', input_shape=(28,28,1)))
model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))

model1.add(Flatten())

model1.add(Dense(128,activation='relu'))
model1.add(Dense(10,activation='softmax'))
```

```python
model1.summary()
```

# Padding with Strides

```python
model2 = Sequential()

model2.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model2.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model2.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model2.add(Flatten())

model2.add(Dense(128,activation='relu'))
model2.add(Dense(10,activation='softmax'))
```

```python
model2.summary()
```

# No padding

```python
model3 = Sequential()

model3.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))
model3.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model3.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

model3.add(Flatten())

model3.add(Dense(128,activation='relu'))
model3.add(Dense(10,activation='softmax'))
```

```python
model3.summary()
```

# Strides

```python
model4 = Sequential()

model4.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model4.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(2,2), activation='relu'))
model4.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(2,2), activation='relu'))

model4.add(Flatten())

model4.add(Dense(128,activation='relu'))
model4.add(Dense(10,activation='softmax'))
```

```python
model4.summary()
```
